## 3. Data Validation

This section checks whether the dataset contains the required columns, unique respondent IDs, and valid numerical values before cleaning.


```python

# Check if all required columns exist

required_columns = [
    "respondent_id",
    "doomscroller",
    "sleep_hours_per_night",
    "bedtime_screen_time_minutes",
    "sleep_latency_minutes",
    "avg_doomscroll_session_minutes",
    "country_region",
    "sleep_quality_category"
]

missing_columns = [
    col for col in required_columns
    if col not in df.columns
]

print("Missing required columns:", missing_columns)
```

```python

# Check for duplicate respondent IDs

print("Duplicate respondent IDs:",
      df["respondent_id"].duplicated().sum())
```

```python

# Check for invalid negative values

print("Negative sleep hours:",
      (df["sleep_hours_per_night"] < 0).sum())

print("Negative bedtime screen time:",
      (df["bedtime_screen_time_minutes"] < 0).sum())

print("Negative sleep latency:",
      (df["sleep_latency_minutes"] < 0).sum())

print("Negative doomscroll session duration:",
      (df["avg_doomscroll_session_minutes"] < 0).sum())
```

```python
# Additional Section 3 checks: missing values, category consistency, and potential outliers

print("\n--- Missing Values Before Cleaning ---")
missing_before = df.isnull().sum()
print(missing_before[missing_before > 0])

print("\n--- Duplicate Row Check ---")
print("Duplicate rows:", df.duplicated().sum())

# Check negative values across available numeric lifestyle and sleep variables.
numeric_check_columns = [
    "sleep_hours_per_night",
    "bedtime_screen_time_minutes",
    "sleep_latency_minutes",
    "avg_doomscroll_session_minutes",
    "caffeine_intake_mg_per_day",
    "exercise_minutes_per_day"
]
print("\n--- Negative Value Checks ---")
for col in numeric_check_columns:
    if col in df.columns:
        print(f"{col}: {(df[col] < 0).sum()}")

# Review categorical values and detect leading/trailing spaces.
categorical_check_columns = [
    "doomscroller",
    "sleep_quality_category",
    "country_region",
    "occupation_status",
    "primary_device_used_at_night",
    "bedtime_routine_type"
]
print("\n--- Category and Whitespace Checks ---")
for col in categorical_check_columns:
    if col in df.columns:
        print(f"\n{col} values:")
        print(df[col].value_counts(dropna=False))
        if df[col].dtype == "object":
            values = df[col].dropna().astype(str)
            extra_spaces = (values != values.str.strip()).sum()
            print("Values with leading/trailing spaces:", extra_spaces)

# Flag possible outliers using the IQR rule; do not automatically remove them.
outlier_check_columns = [
    "sleep_hours_per_night",
    "bedtime_screen_time_minutes",
    "sleep_latency_minutes",
    "avg_doomscroll_session_minutes",
    "caffeine_intake_mg_per_day",
    "exercise_minutes_per_day"
]
outlier_results = []
for col in outlier_check_columns:
    if col in df.columns:
        values = df[col].dropna()
        if not values.empty:
            q1 = values.quantile(0.25)
            q3 = values.quantile(0.75)
            iqr = q3 - q1
            lower_bound = q1 - 1.5 * iqr
            upper_bound = q3 + 1.5 * iqr
            count = int(((values < lower_bound) | (values > upper_bound)).sum())
            outlier_results.append({
                "Variable": col,
                "Lower Bound": lower_bound,
                "Upper Bound": upper_bound,
                "Potential Outliers": count
            })

print("\n--- Potential Outliers (IQR Rule) ---")
display(pd.DataFrame(outlier_results))
print("Potential outliers are flagged for review and are not automatically removed.")
```


### Validation Checks

The dataset was validated by checking the presence of required columns, duplicate respondent identifiers and rows, negative numerical values, missing values, categorical value consistency, and potential outliers using the IQR rule. Potential outliers are flagged for review and are not automatically removed. These checks help assess whether the dataset is suitable for cleaning and further analysis.



## 4. Data Cleaning

This section prepares the dataset for analysis by creating a copy of the original data, checking missing values, and applying appropriate cleaning methods. The original dataset remains unchanged.


```python

# Create a copy of the original dataset

df_clean = df.copy()

print("Original dataset:", df.shape)
print("Working copy:", df_clean.shape)
```

```python

# Show columns with missing values

missing_values = df_clean.isnull().sum()

print(missing_values[missing_values > 0])
```

```python

# Fill missing numerical lifestyle values using the median

df_clean["caffeine_intake_mg_per_day"] = (
    df_clean["caffeine_intake_mg_per_day"].fillna(
        df_clean["caffeine_intake_mg_per_day"].median()
    )
)

df_clean["exercise_minutes_per_day"] = (
    df_clean["exercise_minutes_per_day"].fillna(
        df_clean["exercise_minutes_per_day"].median()
    )
)

print("Numerical lifestyle columns cleaned!")
```

```python

# Fill missing categorical values with Unknown

categorical_columns = [
    "occupation_status",
    "primary_device_used_at_night",
    "bedtime_routine_type"
]

for col in categorical_columns:
    df_clean[col] = df_clean[col].fillna("Unknown")

print("Categorical columns cleaned!")
```

```python

# Check remaining missing values

print("Missing values after cleaning:")

print(
    df_clean.isnull().sum()[
        df_clean.isnull().sum() > 0
    ]
)
```


### Missing Value Treatment

Missing numerical values in caffeine intake and daily exercise were replaced using median imputation. Missing categorical values in occupation status, primary device used at night, and bedtime routine were labelled as "Unknown".

The 40 missing sleep quality scores were retained to avoid introducing artificial outcome values. The cleaning operations were performed on a working copy, preserving the original dataset.


## 6. Data Preparation and Transformations

This section prepares the cleaned dataset for the three visualizations proposed in Deliverable 1:

1. Average Sleep Hours by Doomscrolling Status
2. Bedtime Screen Time vs. Sleep Latency
3. Short and Long Doomscrolling Sessions by Country/Region


```python

# Chart 1: Average sleep hours by doomscrolling status

sleep_summary = (
    df_clean.groupby("doomscroller")["sleep_hours_per_night"]
    .mean()
    .reset_index()
)

sleep_summary
```

```python

# Chart 2: Bedtime screen time vs. sleep latency

scatter_data = df_clean[
    [
        "bedtime_screen_time_minutes",
        "sleep_latency_minutes",
        "sleep_quality_category"
    ]
].dropna()

scatter_data.head()
```

```python

# Chart 3: Classify short and long doomscrolling sessions

SESSION_THRESHOLD = 15

# Keep missing durations unclassified
df_clean["session_category"] = np.where(
    df_clean["avg_doomscroll_session_minutes"].isna(),
    "Unknown",
    np.where(
        df_clean["avg_doomscroll_session_minutes"] <= SESSION_THRESHOLD,
        "Short",
        "Long"
    )
)

# Count session categories by region
region_summary = pd.crosstab(
    df_clean["country_region"],
    df_clean["session_category"]
)

region_summary
```


## 7. Data Export

This section exports the cleaned dataset and three visualization-ready tables as CSV files. The exported files allow the analysis to be reused and reproduced without modifying the original dataset.


```python

# Create a folder for exported files

from pathlib import Path

output_folder = Path("output")
output_folder.mkdir(exist_ok=True)

print("Output folder created successfully!")
```

```python

# Export the cleaned dataset
df_clean.to_csv(
    output_folder / "cleaned_sleep_data.csv",
    index=False
)

# Export Chart 1 data
sleep_summary.to_csv(
    output_folder / "sleep_summary.csv",
    index=False
)

# Export Chart 2 data
scatter_data.to_csv(
    output_folder / "scatter_data.csv",
    index=False
)

# Export Chart 3 data
region_summary.to_csv(
    output_folder / "region_summary.csv"
)

print("All CSV files exported successfully!")
```
